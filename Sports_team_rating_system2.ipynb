```python
# ============================================================
# SPORTS TEAM RATING SYSTEM (B8)
# SIMPLE USER INPUT VERSION
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Automatically create 5 teams
teams = ["Team A", "Team B", "Team C", "Team D", "Team E"]

print("=" * 55)
print("       SPORTS TEAM RATING SYSTEM (B8)")
print("=" * 55)

print("""
Enter the point differences for the matches.

There are 10 matches between 5 teams.
Enter 10 values separated by commas.

Example:
5, -3, 7, 2, 4, -6, 3, 5, -2, 4
""")

# User input
values = input("Enter 10 match point differences: ")

# Convert input into numbers
values = [float(x.strip()) for x in values.split(",")]

# Check input
if len(values) != 10:
    print("Please enter exactly 10 values.")
else:

    # --------------------------------------------------------
    # CREATE POINT DIFFERENCE MATRIX
    # --------------------------------------------------------

    matrix = np.zeros((5, 5))

    matches = [
        (0, 1),  # Team A vs Team B
        (0, 2),  # Team A vs Team C
        (0, 3),  # Team A vs Team D
        (0, 4),  # Team A vs Team E
        (1, 2),  # Team B vs Team C
        (1, 3),  # Team B vs Team D
        (1, 4),  # Team B vs Team E
        (2, 3),  # Team C vs Team D
        (2, 4),  # Team C vs Team E
        (3, 4)   # Team D vs Team E
    ]

    for value, (i, j) in zip(values, matches):
        matrix[i][j] = value
        matrix[j][i] = -value

    print("\nPOINT DIFFERENCE MATRIX:")
    print(pd.DataFrame(matrix, index=teams, columns=teams))

    # --------------------------------------------------------
    # FIND DOMINANT EIGENVALUE AND EIGENVECTOR
    # --------------------------------------------------------

    eigenvalues, eigenvectors = np.linalg.eig(matrix)

    # Find eigenvalue with largest absolute value
    index = np.argmax(np.abs(eigenvalues))

    dominant_eigenvalue = eigenvalues[index]
    dominant_eigenvector = np.real(eigenvectors[:, index])

    # Make ratings positive
    dominant_eigenvector = np.abs(dominant_eigenvector)

    # Normalize ratings
    ratings = dominant_eigenvector / np.sum(dominant_eigenvector)

    # Convert into percentage
    ratings_percentage = ratings * 100

    # --------------------------------------------------------
    # CREATE RESULT TABLE
    # --------------------------------------------------------

    result = pd.DataFrame({
        "Team": teams,
        "Rating": ratings_percentage
    })

    result = result.sort_values(
        by="Rating",
        ascending=False
    ).reset_index(drop=True)

    result["Rank"] = range(1, 6)

    result = result[["Rank", "Team", "Rating"]]

    # --------------------------------------------------------
    # DISPLAY RESULTS
    # --------------------------------------------------------

    print("\n" + "=" * 55)
    print("              TEAM RATINGS")
    print("=" * 55)

    print(result.to_string(
        index=False,
        formatters={"Rating": "{:.2f}".format}
    ))

    print("\nDominant Eigenvalue:",
          round(np.real(dominant_eigenvalue), 4))

    print("\nWinner:", result.iloc[0]["Team"])

    # --------------------------------------------------------
    # BAR CHART
    # --------------------------------------------------------

    plt.figure(figsize=(9, 5))

    plt.bar(
        result["Team"],
        result["Rating"]
    )

    plt.xlabel("Teams")
    plt.ylabel("Rating (%)")
    plt.title("Sports Team Rating System")

    # Display values above bars
    for i, value in enumerate(result["Rating"]):
        plt.text(
            i,
            value + 0.5,
            f"{value:.2f}%",
            ha="center"
        )

    plt.ylim(0, max(result["Rating"]) + 10)
    plt.grid(axis="y", linestyle="--", alpha=0.5)

    plt.show()

    print("\nProject completed successfully!")
```
